# deployment guide

Yes, it is absolutely possible! If your original source code repository no longer exists, you can still use your local computer to build the site and push only the dist folder to a brand-new, empty GitHub repository.

Since you only want the build files on GitHub, we can configure a special git setup inside your project. Here is how to set it up so you can deploy with a single command.

Step 1: Create your new empty GitHub Repository
Go to GitHub and create a new repository (e.g., my-angular-site).

Do not initialize it with a README, .gitignore, or license. Keep it completely empty.

Copy the repository URL (e.g., https://github.com/username/my-angular-site.git).

Step 2: Install the deployment tool locally
Since your original GitHub tracking is gone, we will use angular-cli-ghpages to handle the heavy lifting. It will automatically create a hidden tracking system just for your build files and push them to your new repo.

Open your terminal in your project's root folder and run:

Bash
npm install angular-cli-ghpages --save-dev
Step 3: Configure the deployment command
Open your project's package.json file. Look for the "scripts" section, and add a custom deploy script that points exclusively to your new GitHub repository.

Add this line inside "scripts":

JSON
"scripts": {
  ...
  "deploy": "ng build --configuration production --base-href \"https://<your-username>.github.io/<your-repo-name>/\" && npx angular-cli-ghpages --dir=dist/<your-project-name> --repo=https://github.com/<your-username>/<your-repo-name>.git --branch=main"
}
⚠️ Important adjustments to make in that line:

Replace <your-username> with your GitHub username.

Replace <your-repo-name> with your new empty repository name.

Replace <your-project-name> with the actual folder name generated inside dist/ when you build (check your angular.json under outputPath if you aren't sure).

Note: We use --branch=main so it hosts out of your primary branch instead of creating a secondary gh-pages branch.

Step 4: Run the deployment
Now, whenever you want to update your live site, you just need to run one command in your terminal:

Bash
npm run deploy
What happens when you run this?
Angular compiles your application into pure static files inside the dist/ folder.

The deployment tool creates a temporary, isolated Git instance inside that dist/ folder.

It connects exclusively to the new GitHub URL you provided, completely ignoring your local project's Git status.

It force-pushes those files directly to the main branch of your empty GitHub repository.

If you refresh your GitHub repository page, you will see it is no longer empty—it will contain only files like index.html, main.js, and styles.css.
